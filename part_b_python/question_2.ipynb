- Write	a	short	data-quality	check	on	the	payments	file:	flag	duplicate
payment_ids,	payments	whose	amount	does	not	match	the	rental’s	monthly_rent,
and	payments	linked	to	a	rental_id	that	does	not	exist	in	rentals.csv.	Show	counts and	a	few	example	rows	for	each	issue.

```python
import pandas as pd

payments = pd.read_csv("../data/payments.csv", parse_dates=["payment_date"])
rentals = pd.read_csv("../data/rentals.csv", parse_dates=["start_date", "end_date"])
```

```python
# 1. duplicate payment ids
dup = payments[payments["payment_id"].duplicated()]
print("duplicates payment_id count:", len(dup))
display(dup.head())
```

```python
# 2. payments(amount) not match rentals(monthly_rent)
m = payments.merge(rentals[["rental_id", "monthly_rent"]], on="rental_id")
mismatch = m[m["amount"].round(2) != m["monthly_rent"].round(2)]
print("count of payments(amount) does not match rental(monthly_rent):", len(mismatch))
display(mismatch[["payment_id", "rental_id", "amount", "monthly_rent"]].head())
```

```python
# 3. payments(rental_id) not in rentals(rental_id)
orphan = payments[~payments["rental_id"].isin(rentals["rental_id"])]
print("count of payments(rental_id) not in rentals(renatal_id):", len(orphan))
display(orphan.head())
```

- there is none rows in all three data quality checks so we are getting 0 means data has no errors
